# Business Analysis

```python
df.groupby('Category')[['Revenue', 'Profit']].sum().sort_values(
    by='Revenue', ascending=False
)
```

```python
df.groupby('Region')[['Revenue', 'Profit']].sum().sort_values(
    by='Revenue', ascending=False
)
```

```python
df.groupby('Product_Name')['Revenue'].sum().sort_values(
    ascending=False
).head(10)
```
